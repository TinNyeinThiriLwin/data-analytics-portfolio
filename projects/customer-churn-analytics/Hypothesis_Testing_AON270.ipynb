```python
import pandas as pd
from scipy.stats import chi2_contingency
from statsmodels.stats.proportion import proportions_ztest
```

```python
data = {
    '<270': {'churned': 37908, 'not_churned': 233956},
    '>270': {'churned': 6215942, 'not_churned': 19988579}
} 
```

```python
# for chi-square
contingency_table = [
    [data['<270']['churned'], data['<270']['not_churned']],
    [data['>270']['churned'], data['>270']['not_churned']]
]
```

```python
# Chi-square test
chi2, p_value, dof, expected = chi2_contingency(contingency_table)

print("Chi-square test:")
print(f"Chi2 Statistic = {chi2:.4f}")
print(f"P-value = {p_value:.4f}")
```

```python
# Number of successes (churned) and observations (total) in each group
successes = [data['<270']['churned'], data['>270']['churned']]
nobs = [
    data['<270']['churned'] + data['<270']['not_churned'],
    data['>270']['churned'] + data['>270']['not_churned']
]
```

```python

# Two-proportion z-test
z_stat, p_val = proportions_ztest(successes, nobs)

print("\nTwo-Proportion Z-Test:")
print(f"Z-statistic = {z_stat:.4f}")
print(f"P-value = {p_val:.4f}")
```

(p-value)0.0000 < 0.05: Statistically significant difference in churn rates between <270 and >270.
